```python
import urllib.request, json
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
campaignId = 'd02d863f2cd2'
raw_data= json.loads(urllib.request.urlopen('http://h2615096.stratoserver.net/api/work/%s' % campaignId).read())
```

# Group A: Noto Emoji

```python
groupA = [row for row in raw_data if 'code' in row['items'][0]['choices'][0]]
groupA = [dict(item, worker=row['workerId']) for row in groupA for item in row['items']]
statsA = {}
for item in groupA:
    file = item['file']
    if file not in statsA:
        statsA[file] = {}
    code = '%0x' % item['choices'][item['selected']]['code']
    if code not in statsA[file]:
        statsA[file][code] = 0
    statsA[file][code] += 1
        
statsA
```
