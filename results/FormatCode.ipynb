```python
def dont_split(c):
    return c.isnumeric() or c.isalpha() or c=='.'
def formatNumbers(name, buf, w=WIDTH):
    buf = f"{name}=" + buf
    while len(buf)>0:
        v = w
        if len(buf)>=v and dont_split(buf[v]):
            while dont_split(buf[v-1]):
                v = v - 1
        print(buf[0:v], end="")
        buf = buf[v:]
        if len(buf)>0:
            print("")
            buf = "  " + buf
```

```python
def formatString(name, buf, w=WIDTH, concat_op='+'):
    buf = f"{name}='" + buf
    while len(buf)>0:
        v = w - 2
        print(buf[0:v], end="")
        buf = buf[v:]
        if len(buf)>1:
            print("'" + concat_op)
            buf = "  '" + buf
        else:
            print(buf + "'")
            buf = ""
```
