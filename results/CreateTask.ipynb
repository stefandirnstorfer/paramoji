# Create task

```python
N=50
```

```python
I = np.random.choice(range(len(data)), N)
selection=[data.iloc[i] for i in I]
```

```python
workA=[{
        "file": row['image'].replace('../input/',''),
    } for row in selection]
```

```python
workB = np.array([
    np.round(100*np.random.permutation(np.arange(0, N)/(N-1)))
    for _ in range(5)
]).T.tolist()
workB=[{"paramoji":x} for x in workB]
```

```python
id = "classify-"+datetime.datetime.now().isoformat()
print(f"Writing task definition for {id}")
with open("gen/work.json", "w") as file:
    json.dump({
        "id": id,
        "A": workA,
        "B": workB
    }, file, indent=2)
```
